```python
import scipy.optimize
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 
import pulp 
```

```python
A= [[1,1],[2,1]]
C=[-3,-2]
b=[4,5]
x_1b=(0,None)
x_2b=(0,None)
```

```python
res=scipy.optimize.linprog(C, A_ub=A, b_ub=b, bounds=[x_1b,x_2b], method='Simplex')
```

```python
if res.success:
    print("Optimal solution:", res.x)
    print("Optimal value:", -res.fun)
else:
    print("Optimization failed:", res.message)
```

```python
#Question 2 
problem=pulp.LpProblem("Maximize",pulp.LpMaximize)
x1=problem.add_variable("x1",lowBound=0)
x2=problem.add_variable("x2",lowBound=0)
```

```python

problem += 3*x1 + 2*x2, "Objective"
```

```python
problem += x1 + x2 <= 4, "Constraint 1"
problem += 2*x1 + x2 <= 5, "Constraint 2"
```

```python
problem.solve()
```
